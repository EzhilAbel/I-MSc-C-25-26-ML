```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Generate synthetic dataset
X, y = make_classification(n_samples=300,
                           n_features=2,
                           n_classes=2,
                           n_informative=2,
                           random_state=42)

# BUG 1: Wrong test size value
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=30, random_state=42)

# BUG 2: Incorrect model initialization parameter
model = LogisticRegression(max_iter=10)

# BUG 3: Wrong variable used for training
model.fit(X_test, y_test)

# BUG 4: Wrong prediction variable
y_pred = model.predict(X_train)

# BUG 5: Accuracy calculated using mismatched variables
acc = accuracy_score(y_test, y_pred)
print("Accuracy:", acc)

# Visualization
plt.scatter(X[:,0], X[:,1], c=y, cmap='coolwarm')
plt.title("Decision Boundary")

# BUG 6: Incorrect meshgrid creation
x_min, x_max = X[:,0].min() - 1, X[:,0].max() + 1
y_min, y_max = X[:,1].min() - 1, X[:,1].max() + 1
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 100),
                     np.linspace(y_min, y_max, 100))

# BUG 7: Wrong reshape dimension
Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(50, 50)

plt.contourf(xx, yy, Z, alpha=0.3, cmap='coolwarm')
plt.show()
```
